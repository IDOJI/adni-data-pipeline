```
import pandas as pd
```

```
df_iqms = pd.read_csv('all_IQMs_with_QC_flags_clean.tsv', sep="\t")
```

```
%matplotlib inline
```

```
df_iqms['site'] = df_iqms['participant_id'].str.replace('sub-ADNI','').astype(str).str[:3]
```

```
df_iqms.columns
```

```
t1_iqms = ['t1_Entropy Focus Criterion (EFC)',
       't1_Coefficient of Joint Variation (CJV)',
       't1_Full Width Half Max (FWHM (mm))',
       't1_Contrast to Noise Ratio (CNR)',
       't1_Signal to Noise Ratio (SNR)',
       't1_Tissue Prob Map (TPM) Overlap CSF',
       't1_TPM Overlap GM', 't1_TPM Overlap WM']
```

```
bold_iqms = ['bold_Entropy Focus Criterion (EFC)',
       'bold_Full Width Half Max (FWHM (mm))',
       'bold_AFNIs outlier ratio (AOR)',
       'bold_AFNIs mean quality index (AQI)',
       'bold_Temp deriv RMS var over vox (DVARS)',
       'bold_Framewise disp (meanFD)',
       'bold_Signal to Noise Ratio (SNR)',
       'bold_Temporal SNR (TSNR)']
```

```
import joypy
```

```
import matplotlib.pyplot as plt
import numpy as np
```

```
#pd.set_option('display.max_rows', 500)
#pd.set_option('display.max_columns', 500)
#pd.set_option('display.width', 1000)
def color_gradient(x=0.0, start=(0, 0, 0), stop=(1, 1, 1)):
    r = np.interp(x, [0, 1], [start[0], stop[0]])
    g = np.interp(x, [0, 1], [start[1], stop[1]])
    b = np.interp(x, [0, 1], [start[2], stop[2]])
    return (r, g, b)
```

```
for i in t1_iqms:
    plt.figure(dpi=380)
    fig, axes = joypy.joyplot(df_iqms, column=[i], overlap=1.0, by="site", ylim='own', fill=True, figsize=(5,14), legend=False, xlabels=True, ylabels=True, 
                          colormap=lambda x: color_gradient(x, start=(.08, .45, .8),stop=(.8, .34, .44)), alpha=0.6, linewidth=.5, linecolor='w', fade=True);
    j = i.strip('t1_')
    plt.title("MRIQC T1w IQMs - {}".format(j), fontsize=12, color='black', alpha=1);
    plt.rc("font", size=10);
    plt.xlabel("{}".format(j), fontsize=12, color='black', alpha=1);
    plt.ylabel('Site', fontsize=10, color='black', alpha=1);
    plt.savefig("{}.png".format(i), dpi=300, bbox_inches = "tight")
```

```
for i in bold_iqms:
    plt.figure(dpi=380)
    fig, axes = joypy.joyplot(df_iqms, column=[i], overlap=1.0, by="site", ylim='own', fill=True, figsize=(5,14), legend=False, xlabels=True, ylabels=True, 
                          colormap=lambda x: color_gradient(x, start=(.08, .45, .8),stop=(.8, .34, .44)), alpha=0.6, linewidth=.5, linecolor='w', fade=True);
    j = i.strip('bold_')
    plt.title("MRIQC BOLD IQMs - {}".format(j), fontsize=12, color='black', alpha=1);
    plt.rc("font", size=10);
    plt.xlabel("{}".format(j), fontsize=12, color='black', alpha=1);
    plt.ylabel('Site', fontsize=10, color='black', alpha=1);
    plt.savefig("{}.png".format(i), dpi=300, bbox_inches = "tight")
```
